## Tutorial: a two-location quantum piece
Use `|0⟩ = A1`, `|1⟩ = A2`. A Hadamard creates equal amplitudes; Z changes relative phase without immediately changing location probabilities. Applying H again makes phase visible.

For a normalized state `a|0⟩ + b|1⟩`, the probability of A1 is `|a|²`, and of A2 is `|b|²`.

````python
from qiskit.quantum_info import Statevector

def example_piece(phase_flip=False):
    qc = QuantumCircuit(1)
    qc.h(0)
    if phase_flip:
        qc.z(0)
    qc.h(0)
    return Statevector.from_instruction(qc)

for phase in [False, True]:
    print("Phase flip:", phase, "location probabilities:", example_piece(phase).probabilities_dict())
````
